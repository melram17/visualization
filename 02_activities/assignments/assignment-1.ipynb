```python
#Importing libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#tested out different graphs in class (scatter, line) here im showing the line plot
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots (figsize=(5,3))
ax.scatter(x,y)
```

```python
#matplotlib basic figures, testing out histogram as per class challenge
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots (figsize=(5,3))
ax.hist(y)
```

```python
#adding labels
fig, ax = plt.subplots (figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')

#modifying labels
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
#adjusting font
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
#changing location of title, put loc towards the left
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
#stars, yayyy :D
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter( x,
y,
marker='*',
color = "green") #changed the colour from indigo to green
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
y,
marker='*',
color = 'indigo',
linestyle = '--',
linewidth = 2)
fig.show()
```

```python
#appling hex codes
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
fig.show()
```

```python
ax.plot(x,y,marker='*',
markersize = 12,
color = '#7425b9',
linestyle = '--',
linewidth = 2,
markeredgecolor = '#fa9359',
markerfacecolor = '#fa9359' )
```

```python
#adding grids
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
ax.grid(axis = 'y', color = "blue", linewidth = 2, linestyle = '-.')
```
